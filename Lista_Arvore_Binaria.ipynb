# Exercício 1

```python
def criar_arvore(valores):
    
    arvore = {}
    
    for valor in valores:
    
        # Verifica se a árvore está vazia (primeiro elemento)
        if len(arvore) == 0:
            # Cria o nó raiz com ponteiros para esquerda e direita vazios
            arvore[valor] = {"esquerda": None, "direita": None}
            raiz = valor  # Define e guarda quem é a raiz principal da árvore
    
        # Se a árvore já tiver elementos
        else:
            # Reinicia a busca sempre a partir da raiz principal
            atual = raiz 
    
            # Loop para descer na árvore até encontrar a posição correta
            while True:
    
                # Se o valor for menor, vai para a esquerda
                if valor < atual:
    
                    # Se o lado esquerdo estiver vazio, insere o novo nó aqui
                    if arvore[atual]["esquerda"] is None: 
                        arvore[atual]["esquerda"] = valor # Conecta o nó atual ao novo valor
                        arvore[valor] = {"esquerda": None, "direita": None}  # Inicializa a estrutura do novo nó
                        break
    
                    # Se já existir um nó na esquerda, desce para ele e continua o loop
                    else:
                        atual = arvore[atual]["esquerda"]
    
                # Se o valor for maior ou igual, vai para a direita
                else:
    
                    # Se o lado direito estiver vazio, insere o novo nó aqui
                    if arvore[atual]["direita"] is None: 
                        arvore[atual]["direita"] = valor # Conecta o nó atual ao novo valor
                        arvore[valor] = {"esquerda": None, "direita": None}  # Inicializa a estrutura do novo nó
                        break
                    
                    # Se já existir um nó na direita, desce para ele e continua o loop
                    else:
                        atual = arvore[atual]["direita"]
    return arvore, raiz
valores = [53, 30, 72, 14, 39, 61, 84, 9, 23, 34, 49, 79]
arvore, raiz = criar_arvore(valores)

print("Árvore criada:")
print(arvore)
print(f"Raiz: {raiz}(primeiro valor do vetor)")
```

# Exercício 3

```python
def procurar_valor_arvore(arvore, valor_procurado):
    
    caminho = []
    quant_nos = 0
    
    for i in arvore:
        raiz = i
        break
        
    atual = raiz # Muda para a raiz da arvore
    
    encontrou = False
    
    while atual is not None:
        # Adiciona ao caminho
        caminho.append(atual)
        quant_nos += 1
    
        # Verifica se o valor já é o procurado
        if valor_procurado==atual:
            encontrou = True
            break
    
        # Verifica se o valor é menor que o nó atual, para descer na esquerda da arvore
        elif valor_procurado < atual:
            atual = arvore[atual]["esquerda"]
            
        # Verifica se o valor é maior que o nó atual, para descer na direita da arvore
        else:
            atual = arvore[atual]["direita"]
    
    if encontrou:
        print("\nValor encontrado")
    else:
        print("\nValor não encontrado")
    
    print(f"Caminho do valor: {caminho}")

    print(f"Nós visitados: {quant_nos}")
    
```

# Exercício 4

```python
def travessia_pre_ordem(arvore):
    pre_ordem = [] # Cria a lista 

    # Pega a raiz
    for i in arvore:
        raiz = i
        break
        
    # Pega o valor da raiz, primeiro nó
    pilha = [raiz] # 53
    
    while len(pilha) > 0: # Enquanto a pilha tiver algum valor, ou seja, nó
    
        atual = pilha.pop() # Pega o valor do nó para verificar agora e tira da pilha
    
        pre_ordem.append(atual) # Adiciona o no caminho o nó verificado
    
        if arvore[atual]["direita"] is not None: # Verifica se há algum valor na direita da arvore
    
            pilha.append(arvore[atual]["direita"]) # Se tiver, ele adiciona esse valor na pilha, a fim de verficar depois
    
        if arvore[atual]["esquerda"] is not None: # Verifica se há algum valor na esquerda da arvore
    
            pilha.append(arvore[atual]["esquerda"]) # Se tiver, ele adiciona esse valor na pilha, a fim de verficar depois

        # Com isso, o ultimo valor na pilha é sempre a esquerda pq é o ultimo adicionado, sendo o proximo verificado, seguindo o contexto: raiz, esquerda, direita

    print([raiz])
    print("Travessia pré-ordem:")
    print(pre_ordem)
```

```python
travessia_pre_ordem(arvore)

print("a pilha serve para garantir a sequencia da pre ordem, e garantir que nao tenha nenhum nó que nao foi visitado")
```

```python
def travessia_em_ordem(arvore):
    em_ordem = [] # Cria a lista 
    pilha = []

    # Pega a raiz da arvore(53)
    for i in arvore:
        raiz = i
        break

    atual = raiz # Pega o valor atual a ser visto(53), começando da raiz da arvore

    while atual is not None or len(pilha) > 0: # Verifica se a pilha é maior que 0, ou se ainda há raiz(atual) para verificar

        while atual is not None: # Enquanto o atual não for nulo, continua descendo na arvore
    
            pilha.append(atual) # Adiciona na pilha o atual para lembrar dele depois
            
            # print(pilha)
    
            atual = arvore[atual]["esquerda"] # Desce para o filho da esquerda até encontrar o ultimo valor nulo
    
        atual = pilha.pop() # Remove e pega o ultimo valor que foi guardado na pilha
    
        em_ordem.append(atual) # Adiciona na lista esse valor que acabou de ser retirado da pilha
    
        atual = arvore[atual]["direita"] # Como já olhou a esquerda e o nó atual, ele cai para o filho da direita
        
        # print("Atual: ", atual)
        # print("Ordem: ", em_ordem)
        
    print("Travessia em ordem:")
    print(em_ordem)
```

```python
travessia_em_ordem(arvore)
print("o resultado aparece ordenado porque a travessia acontece com todos os nós a esquerda que são os menores, somente após ele vai voltando e verificando os demais nós")
print("a diferença entre pilha e recursão é que em pilha fazemos o controle da ordem e volta e em recursao ela mesma faz essa ordenação guardando os passos.")
```

```python
def travessia_pos_ordem(arvore):
    pos_ordem = [] # Cria a lista

    # Pega a raiz da arvore(53)
    for i in arvore:
        raiz = i
        break
        
    pilha1 = [raiz] # Cria uma pilha que pega o valor da raiz e seus ramos
    
    pilha2 = [] # Cria uma outra pilha vazia
    
    while len(pilha1) > 0: # Enquanto a pilha for maior que 0, ele continua o código
    
        atual = pilha1.pop() # Tira o ultimo valor da pilha e adiciona no atual
    
        pilha2.append(atual) # Adiciona o atual na segunda pilha

        # print("Pilha 1: ", pilha1)
        # print("Pilha 2: ", pilha2)
    
        if arvore[atual]["esquerda"] is not None: # Se existir algo na esquerda do nó atual
    
            pilha1.append(arvore[atual]["esquerda"]) # Adiciona esse valor da esquerda na pilha 1
    
        if arvore[atual]["direita"] is not None: # Se existir algo na direita do nó atual
    
            pilha1.append(arvore[atual]["direita"]) # Adiciona esse valor da direita na pilha 1

        # print("Atual: ", atual)
    
    while len(pilha2) > 0: # Se a pilha 2 tiver algum valor
    
        pos_ordem.append(pilha2.pop()) # Adiciona o ultimo valor da pilha 2 na lista pos_ordem
        # Ele cria a pos ordem ao contrario, por isso a pilha2 utilizando pop
    
    print("Travessia pós-ordem:") 
    print(pos_ordem) 
```

```python
travessia_pos_ordem(arvore)
print("foi necessario ultilizar duas pilhas pois uma foi usada para registrar a ordem percorrida e a outra foi ultilizada para inverter a posição")
print("no pre ordem ele percorre e começa pela raiz e em ordem ele desce todos pra esquerda e começa do fim visitando os nós acima, a raiz se torna o ponto central")
```

# Exercício 7


```python
def altura_arvore(arvore):
    # Raiz
    for i in arvore:
        raiz = i
        break

    #Inicializa as variaveis para o resultado
    caminhos = [] # Lista para conter os caminhos

    pilha = [(raiz, [raiz])] # Armazena não apenas a raiz, mas uma lista para ter o caminho de cada nó

    while len(pilha) > 0:

        
        atual, caminho = pilha.pop() # Pega o valor do nó e do caminho em uma tupla

        # Se não tem ninguém na esquerda E nem na direita, salvamos esse caminho pronto, só precisamos dos caminhos finais, não de todos
        if arvore[atual]["esquerda"] is None and arvore[atual]["direita"] is None:
            caminhos.append(caminho)


        # Percorre a arvore
        if arvore[atual]["esquerda"] is not None: # Verifica se tem nós a esquerda
            filho_esq = arvore[atual]["esquerda"] # Cria uma variavel com o valor do nó a esquerda
            pilha.append( (filho_esq, (caminho + [filho_esq])) ) # Adiciona na pilha o nó que ira adentrar agora, e adiciona o caminho com o novo nó adicionado

        if arvore[atual]["direita"] is not None: # Verifica se tem nós a esquerda
            filho_dir = arvore[atual]["direita"] # Cria uma variavel com o valor do nó a esquerda
            pilha.append( (filho_dir, (caminho + [filho_dir])) ) # Adiciona na pilha o nó que ira adentrar agora, e adiciona o caminho com o novo nó adicionado


    maior_caminho = max(caminhos, key=len) # Mostra o maior em questao de len, ou seja, tamanho do vetor
    altura = len(maior_caminho) # Pega o tamano do vetor

    print(maior_caminho)
    print(altura)
```

```python
def contar_nos(arvore):
    # Pega a raiz da arvore
    for i in arvore:
        raiz = i
        break
    
    # Inicializa a contagem
    quant_nos = 0
    quant_nos_duplos = 0
    quant_nos_folha = 0
    quant_nos_unicos = 0

    # Pega o valor da raiz, primeiro nó
    pilha = [raiz] # 53
    
    # Enquanto houver algum nó para verificar(pilha não for vazia)
    while len(pilha) > 0:

        # Transforma o atual no nó que veremos e retira da pilha, deixando vazia
        atual = pilha.pop()

        # Como achamos um nó, aumenta um no total
        quant_nos += 1
    


        # Desce na arvore
        # Verifica se tem um nó filho a direita, se tiver, ele cai pra ela
        if arvore[atual]["direita"] is not None:
            pilha.append(arvore[atual]["direita"]) # Adiciona o nó a direita na pilha para verificar no while
    
        # Verifica se tem um nó filho a direita, se tiver, ele cai pra ela
        if arvore[atual]["esquerda"] is not None:
            pilha.append(arvore[atual]["esquerda"]) # Adiciona o nó a esquerda na pilha para verificar no while



        # Verifica a contagem dos nós
        if arvore[atual]["esquerda"] is not None and arvore[atual]["direita"] is not None: # Verifica se tem dois filhos
            quant_nos_duplos += 1

        if (arvore[atual]["esquerda"] is not None and arvore[atual]["direita"] is None) or (arvore[atual]["esquerda"] is None and arvore[atual]["direita"] is not None) : # Verifica se tem um filho
            quant_nos_unicos += 1

        if arvore[atual]["esquerda"] is None and arvore[atual]["direita"] is None:
            quant_nos_folha += 1

    
    print(quant_nos)
    print(quant_nos_unicos)
    print(quant_nos_duplos)
    print(quant_nos_folha)
```

```python
def remover_valor(arvore, valor):

    pai = None
    atual = 53  # raiz

    # Procura o nó e seu pai
    while atual is not None and atual != valor:

        pai = atual

        if valor < atual:
            atual = arvore[atual]["esquerda"]
        else:
            atual = arvore[atual]["direita"]

    # Valor não encontrado
    if atual is None:
        print("Valor não encontrado.")
        return

    esquerda = arvore[atual]["esquerda"]
    direita = arvore[atual]["direita"]

    # CASO 1 - Nó folha
    if esquerda is None and direita is None:

        if arvore[pai]["esquerda"] == valor:
            arvore[pai]["esquerda"] = None
        else:
            arvore[pai]["direita"] = None

        del arvore[valor]

        print(f"{valor} removido (nó folha).")

    # CASO 2 - Nó com um filho
    elif esquerda is None or direita is None:

        if esquerda is not None:
            filho = esquerda
        else:
            filho = direita

        if arvore[pai]["esquerda"] == valor:
            arvore[pai]["esquerda"] = filho
        else:
            arvore[pai]["direita"] = filho

        del arvore[valor]

        print(f"{valor} removido (um filho).")

    # CASO 3 - Nó com dois filhos
    else:

        sucessor_pai = atual
        sucessor = direita

        while arvore[sucessor]["esquerda"] is not None:
            sucessor_pai = sucessor
            sucessor = arvore[sucessor]["esquerda"]

        print(f"Sucessor encontrado: {sucessor}")

        # Liga o sucessor ao filho esquerdo do nó removido
        arvore[sucessor]["esquerda"] = esquerda

        # Ajusta a posição antiga do sucessor
        if sucessor_pai != atual:
            arvore[sucessor_pai]["esquerda"] = arvore[sucessor]["direita"]
            arvore[sucessor]["direita"] = direita

        # Liga o pai ao sucessor
        if pai is None:
            pass
        elif arvore[pai]["esquerda"] == valor:
            arvore[pai]["esquerda"] = sucessor
        else:
            arvore[pai]["direita"] = sucessor

        del arvore[valor]

        print(f"{valor} removido (dois filhos).")
```

```python
def buscar_largura_niveis_lista(arvore):
    
    for i in arvore:
        raiz = i
        break
    
    # Não é mais uma pilha, sim uma fila(FIFO, não LIFO)
    fila = [raiz] # Armazena a raiz
    
    nivel_atual = 1

    while len(fila) > 0: # Passamos por todos os nós

        tamanho_do_nivel = len(fila) # Avalia quantos nós este nivel tem, começa apenas com a raiz
        nos_deste_nivel = [] # Inicia a variavel para guardar os valoes
        
        for _ in range(tamanho_do_nivel):
            
            atual = fila.pop(0) # Pop(0) remove o primeiro elemento e armazena o nó
            nos_deste_nivel.append(atual) # Adiciona os nós do nivel
            
            if arvore[atual]["esquerda"] is not None: # Se há algo na esquerda do nó
                fila.append(arvore[atual]["esquerda"]) # Adicione na fila
                
            if arvore[atual]["direita"] is not None: # Se há algo na direita do nó
                fila.append(arvore[atual]["direita"]) # Adicione na fila
        
        print(f"Nivel {nivel_atual}: {nos_deste_nivel}") # Mostra as informações de cada nivel
        
        if len(fila) > 0: # Se a fila ainda tiver numeros
            nivel_atual += 1 # Sobe um nivel

    print(f"\nTotal de níveis na árvore: {nivel_atual}")
```

```python
buscar_largura_niveis_lista(arvore)
```

# Exercício 15

```python
def existe(arvore, valor):
     # Pega a raiz
    for i in arvore:
        raiz = i
        break
    
    pilha = [raiz] # Armazena a raiz

    while len(pilha) > 0:
    
        atual = pilha.pop() # Pega o nó atual e o caminho

        if valor == atual: # Se o valor procurado for encontrado ele usara um return para acabar a função
            return True
            
        elif valor < atual: # Se for menor, tenta cair para a esquerda
            if arvore[atual]["esquerda"] is not None: # Verifica se tem nós a esquerda
                    pilha.append( arvore[atual]["esquerda"] ) # Adiciona na pilha o nó que ira adentrar agora, e adiciona o caminho com o novo nó adicionado
            
        elif valor > atual: # se for maior, tenta cair para a direita
            if arvore[atual]["direita"] is not None: # Verifica se tem nós a esquerda
                pilha.append( arvore[atual]["direita"] ) # Adiciona na pilha o nó que ira adentrar agora, e adiciona o caminho com o novo nó adicionado
  

    return False # Caso ele desça pela arvore toda e não ache o valor(o return), ele mostra que não tem este valor
```

```python
print(existe(arvore, 23))
print(existe(arvore, 100))
print(existe(arvore, 61))
```

# Exercício 16

```python
def somar_nos(arvore):
    # Pega a raiz da arvore
    for i in arvore:
        raiz = i
        break
    
    # Inicializa a contagem
    quant_nos = 0
    total = 0

    # Pega o valor da raiz, primeiro nó
    pilha = [raiz] # 53
    
    # Enquanto houver algum nó para verificar(pilha não for vazia)
    while len(pilha) > 0:

        # Transforma o atual no nó que veremos e retira da pilha, deixando vazia
        atual = pilha.pop()
        total += atual

        # Como achamos um nó, aumenta um no total
        quant_nos += 1
    
        # Desce na arvore
        # Verifica se tem um nó filho a direita, se tiver, ele cai pra ela
        if arvore[atual]["direita"] is not None:
            pilha.append(arvore[atual]["direita"]) # Adiciona o nó a direita na pilha para verificar no while
    
        # Verifica se tem um nó filho a direita, se tiver, ele cai pra ela
        if arvore[atual]["esquerda"] is not None:
            pilha.append(arvore[atual]["esquerda"]) # Adiciona o nó a esquerda na pilha para verificar no while

    print("Quantidade de nós", quant_nos)
    print("Soma dos nós: ", total)
    print("Média da soma dos nos: ", total/quant_nos)
    
```

```python
somar_nos(arvore)
```

# Exercício 17

```python
def gerar_lista_ordenada(arvore):
    lista_ordenada = [] # Cria a lista 
    pilha = []

    # Pega a raiz da arvore(53)
    for i in arvore:
        raiz = i
        break

    atual = raiz # Pega o valor atual a ser visto(53), começando da raiz da arvore


    while atual is not None or len(pilha) > 0: # Verifica se a pilha é maior que 0, ou se ainda há raiz(atual) para verificar

        while atual is not None: # Enquanto o atual não for nulo, continua descendo na arvore
    
            pilha.append(atual) # Adiciona na pilha o atual para lembrar dele depois
                
            atual = arvore[atual]["esquerda"] # Desce para o filho da esquerda até encontrar o ultimo valor nulo
    
        atual = pilha.pop() # Remove e pega o ultimo valor que foi guardado na pilha

    
        lista_ordenada.append(atual) # Adiciona na lista esse valor que acabou de ser retirado da pilha
    
        atual = arvore[atual]["direita"] # Como já olhou a esquerda e o nó atual, ele cai para o filho da direita
        
    print("Travessia em ordem:")
    print(lista_ordenada)
```

```python
gerar_lista_ordenada(arvore)
print("uma ABB gera uma lista ordenada pois sempre o nó a esquerda sera um valor menor e a direita um valor maior")
```

```python
def altura(no):
    # se o nó não existe, a altura é 0
    if no is None:
        return 0

    # calcula a altura do lado esquerdo
    altura_esquerda = altura(arvore[no]["esquerda"])

    # calcula a altura do lado direito
    altura_direita = altura(arvore[no]["direita"])

    # retorna 1 (nó atual) + a maior altura encontrada
    return 1 + max(altura_esquerda, altura_direita)

# define a raiz da árvore
raiz = 53

# calcula a altura da subárvore esquerda da raiz
altura_esquerda = altura(arvore[raiz]["esquerda"])

# calcula a altura da subárvore direita da raiz
altura_direita = altura(arvore[raiz]["direita"])

# mostra as alturas encontradas
print("Altura da esquerda:", altura_esquerda)
print("Altura da direita:", altura_direita)

# calcula a diferença entre as alturas
diferenca = abs(altura_esquerda - altura_direita)

print("Diferença:", diferenca)

# verifica se a árvore está balanceada
# uma árvore balanceada possui diferença de no máximo 1
if diferenca <= 1:
    print("a árvore está balanceada.")
else:
    print("a árvore não está balanceada.")
```
